### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

```python
dataset = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv", sep='\t')
```

### Step 3. Assign it to a variable called chipo.

```python
chipo = dataset.copy()
```

### Step 4. See the first 10 entries

```python
chipo.head(10)
```

### Step 5. Create a histogram of the top 5 items bought

```python
letter_counts = chipo.item_name.value_counts().to_dict()
df = pd.DataFrame.from_dict(letter_counts, orient='index')
df = df[0].sort_values(ascending = True).tail(10)
df.plot(kind='bar')

plt.xlabel('Items')
plt.ylabel('Number of Times Ordered')
plt.title('Most ordered Chipotle\'s Items')

plt.show()
```

### Step 6. Create a scatterplot with the number of items orderered per order price
#### Hint: Price should be in the X-axis and Items ordered in the Y-axis

```python
chipo.item_price = [float(value[1:-1]) for value in chipo.item_price]
```

```python
orders_groupby_id = chipo.groupby('order_id').sum()
plt.scatter(x = orders_groupby_id.item_price, y = orders_groupby_id.quantity, s = 50, c = 'green')

plt.xlabel('order price')
plt.ylabel('Number of Times Ordered')
plt.title('number of items orderered per order price')

plt.show()
```
